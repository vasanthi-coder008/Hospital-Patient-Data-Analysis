```python
import pandas as pd
import numpy as np
from google.colab import files
uploaded = files.upload()
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)
print("Dataset loaded successfully")
print("Shape before cleaning:", df.shape)
print("\nMissing values before cleaning:")
print(df.isnull().sum().sum())
print("\nDuplicate rows before cleaning:")
print(df.duplicated().sum())
df = df.drop_duplicates()
numerical_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include="object").columns
for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())
for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])
print("\nShape after cleaning:", df.shape)
print("\nMissing values after cleaning:")
print(df.isnull().sum().sum())
print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())
print("\nRisk distribution:")
print(df["Risk_Level"].value_counts())
df.to_csv("hospital_patient_dataset_cleaned.csv", index=False)
print("\nDATA CLEANING COMPLETED SUCCESSFULLY")
files.download("hospital_patient_dataset_cleaned.csv")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ============================================================
# GRAPH 2 — TRIAGE LEVEL DISTRIBUTION
# ============================================================

triage_levels = ["1", "2", "3", "4", "5"]
triage_counts = [186, 188, 209, 190, 229]

plt.figure(figsize=(7, 5), facecolor="black")

ax = plt.gca()
ax.set_facecolor("black")

bars = ax.bar(
    triage_levels,
    triage_counts,
    width=0.65
)

plt.title(
    "Triage Level Distribution",
    fontsize=14,
    color="white",
    loc="left",
    pad=15
)

plt.text(
    0,
    240,
    "Distribution of patients across Triage Levels 1 to 5.",
    fontsize=9,
    color="lightgray"
)

plt.ylabel("")
plt.xlabel("")

plt.ylim(0, 260)
plt.yticks(np.arange(0, 261, 65), color="lightgray")
plt.xticks(color="lightgray")

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.25
)

ax.tick_params(axis="both", colors="lightgray")

for spine in ax.spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ============================================================
# GRAPH 1: READMISSION RATE BY AGE GROUP
# ============================================================

age_groups = ["Age < 65", "Age ≥ 65"]
readmission_rates = [21.0, 25.4]

plt.figure(figsize=(7, 5))

bars = plt.bar(
    age_groups,
    readmission_rates,
    width=0.55
)

plt.title("Readmission Rate by Age Group", fontsize=14)
plt.xlabel("")
plt.ylabel("Readmission Rate (%)", fontsize=11)

plt.ylim(0, 30)
plt.yticks(np.arange(0, 31, 5))

# Add values on top of bars
for bar, value in zip(bars, readmission_rates):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.5,
        f"{value:.1f}%",
        ha="center",
        va="bottom",
        fontsize=10
    )

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```
